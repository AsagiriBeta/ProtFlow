# 示例：使用共享工具的新 Notebook

这是一个演示如何使用新的共享工具和标准化初始化的示例 notebook。

**特点：**
- ✅ 自动依赖检查和安装
- ✅ 标准化的环境设置
- ✅ 可重用的代码
- ✅ 更简洁的结构

## 1. 环境初始化

这个单元格替代了之前的多个初始化单元格。

````python
# 环境设置与依赖检查
import sys
from pathlib import Path

# Add protflow to path
project_root = Path.cwd()
while not (project_root / 'src' / 'protflow').exists() and project_root != project_root.parent:
    project_root = project_root.parent

if (project_root / 'src').exists():
    src_dir = str(project_root / 'src')
    if src_dir not in sys.path:
        sys.path.insert(0, src_dir)
    print(f"✓ protflow 路径: {src_dir}")

# Setup environment (automatically checks and installs dependencies)
from protflow.utils.notebook_utils import setup_analysis_notebook

paths = setup_analysis_notebook(work_dir_name='example_runs')

# Common imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Store paths
PROJECT_ROOT = paths['PROJECT_ROOT']
WORK_DIR = paths['WORK_DIR']
DATA_DIR = paths['DATA_DIR']

print(f"\n✓ 初始化完成. 工作目录: {WORK_DIR}")
````

## 2. 使用共享工具

现在可以直接从 `protflow` 导入功能，无需在 notebook 中重新定义。

````python
# 例如：导入序列解析工具
from protflow.utils.seq_parser import parse_genbank, extract_sequences

# 例如：导入 CDS 比较工具
from protflow.core.cds_comparison import compare_annotations

print("✓ 工具导入成功")
````

## 3. 额外依赖

如果需要额外的包，使用共享函数安装：

````python
from protflow.utils.notebook_utils import check_and_install_packages

# 检查并安装额外的包
check_and_install_packages([
    'seaborn',
    ('sklearn', 'scikit-learn')  # (import_name, package_name)
])

import seaborn as sns
print("✓ 额外包已安装")
````

## 4. 你的分析代码

现在可以专注于实际的分析工作...

````python
# 你的代码放在这里
print(f"项目根目录: {PROJECT_ROOT}")
print(f"工作目录: {WORK_DIR}")
print(f"数据目录: {DATA_DIR}")
````
