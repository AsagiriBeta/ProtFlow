## 1. 环境初始化

这个单元格替代了之前的多个初始化单元格。

````python
# 环境设置import sysimport subprocessfrom pathlib import Path# 创建工作目录WORK_DIR = Path('example_runs')WORK_DIR.mkdir(exist_ok=True, parents=True)print(f"✓ 工作目录: {WORK_DIR}")# Helper function to install packages if neededdef install_if_needed(package):    """安装包（如果需要）"""    try:        if isinstance(package, tuple):            import_name, pkg_name = package        else:            import_name = pkg_name = package        __import__(import_name)        print(f"✓ {pkg_name} 已安装")    except ImportError:        print(f"正在安装 {pkg_name}...")        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg_name],                            stdout=subprocess.DEVNULL)        print(f"✓ {pkg_name} 安装完成")# 安装常用包for pkg in ['pandas', 'numpy', 'matplotlib', 'biopython']:    install_if_needed(pkg)print("\n✓ 环境设置完成")
````
